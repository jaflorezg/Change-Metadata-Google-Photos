```python
# Import Libraries
import os
import json
from datetime import datetime
import piexif
```

```python

# Set the folder path where your photos are located
FOLDER_PATH = os.path.expanduser("~/Documents/Photos")
```

```python
#Check the folder path
print(FOLDER_PATH)
```

```python
# Function to fix photo metadata with JSON files
def fix_photos_metadata():
    for filename in os.listdir(FOLDER_PATH):
        if filename.lower().endswith(('.jpg', '.jpeg', '.png')):
            photo_path = os.path.join(FOLDER_PATH, filename)
            json_path = photo_path + ".json"

            if os.path.exists(json_path):
                with open(json_path, 'r') as f:
                    data = json.load(f)
                    # Extract the timestamp from 'photoTakenTime'
                    timestamp = int(data['photoTakenTime']['timestamp'])
                    date_obj = datetime.fromtimestamp(timestamp)
                    exif_date = date_obj.strftime("%Y:%m:%d %H:%M:%S")

                print(f"Processing: {filename} -> Date: {exif_date}")

                try:
                    # 1. Change the date of the system in the file (Mac Finder)
                    os.utime(photo_path, (timestamp, timestamp))

                    # 2. Modify the EXIF metadata (what Google Photos reads)
                    exif_dict = {"Exif": {piexif.ExifIFD.DateTimeOriginal: exif_date}}
                    exif_bytes = piexif.dump(exif_dict)
                    piexif.insert(exif_bytes, photo_path)
                except Exception as e:
                    print(f"Error in {filename}: {e}")

if __name__ == "__main__":
    fix_photos_metadata()
    print("¡Done! Now you can upload the photos and Google will recognize the dates.")
```
